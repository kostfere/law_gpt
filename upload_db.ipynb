```python
import pandas as pd
import openai
import streamlit as st
import openai
import pinecone
from langchain.text_splitter import CharacterTextSplitter
import itertools
from tqdm import tqdm
from utils import upload_embeddings_to_pinecone, batch_embeddings
# from typing import Callable, List, Any


PINECONE_ENVIRONMENT = "gcp-starter"
INDEX_NAME = "law-gpt"
PINECONE_API_KEY = st.secrets["PINECONE_API_KEY"]
OPENAI_API_KEY = st.secrets["OPENAI_API_KEY"]

client = openai.OpenAI(api_key=OPENAI_API_KEY)

df = pd.read_csv("processed_data.csv")
```

```python

text_splitter = CharacterTextSplitter(separator="\n", chunk_size=10000)
# Initialize Pinecone
pinecone.init(api_key=PINECONE_API_KEY, environment=PINECONE_ENVIRONMENT)

# Delete the index if exists
if INDEX_NAME in pinecone.list_indexes():
    pinecone.delete_index(INDEX_NAME)
    print(f"Index '{INDEX_NAME}' has been deleted.")
else:
    print(f"Index '{INDEX_NAME}' does not exist.")

if INDEX_NAME not in pinecone.list_indexes():
    pinecone.create_index(INDEX_NAME, dimension=1536)  # Ensure the dimension is correct
index = pinecone.Index(INDEX_NAME)
upload_embeddings_to_pinecone(df, text_splitter, batch_embeddings,client ,index)
```
